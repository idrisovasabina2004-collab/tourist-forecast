```python
!pip install catboost xgboost lightgbm scikit-learn pandas numpy scipy -q

import time
from datetime import date, timedelta

import numpy as np
import pandas as pd
from catboost import CatBoostRegressor
from xgboost import XGBRegressor
import lightgbm as lgb
from sklearn.metrics import (
    mean_absolute_percentage_error, mean_absolute_error,
    mean_squared_error, r2_score,
)
from sklearn.model_selection import TimeSeriesSplit
from scipy.stats import wilcoxon, ttest_rel

np.random.seed(42)
```

```python
N_ATTRACTIONS = 30
types = ["indoor", "outdoor", "mixed"]
base_range = {
    "indoor":  (500, 30000),
    "outdoor": (1000, 40000),
    "mixed":   (2000, 50000),
}

ATTRACTIONS = {}
for i in range(N_ATTRACTIONS):
    t = np.random.choice(types, p=[0.4, 0.4, 0.2])
    base = int(np.random.uniform(*base_range[t]))
    ATTRACTIONS[f"Attraction_{i:02d}"] = {"base": base, "type": t}

WEATHER_BY_SEASON = {
    "winter": ["Snow", "Cloudy"],
    "spring": ["Cloudy", "Rain", "Clear"],
    "summer": ["Clear", "Cloudy", "Rain"],
    "autumn": ["Cloudy", "Rain"],
}

WEATHER_IMPACT = {
    "outdoor": {"Clear": 1.1, "Cloudy": 0.9, "Rain": 0.6, "Snow": 0.4},
    "indoor":  {"Clear": 1.0, "Cloudy": 1.0, "Rain": 1.05, "Snow": 1.1},
    "mixed":   {"Clear": 1.05, "Cloudy": 0.95, "Rain": 0.8, "Snow": 0.7},
}

HOLIDAYS = [(1, 1), (5, 1), (7, 14), (12, 25), (12, 31)]

def get_season(month):
    if month in (12, 1, 2): return "winter"
    if month in (3, 4, 5):  return "spring"
    if month in (6, 7, 8):  return "summer"
    return "autumn"

def covid_factor(d):
    if date(2020, 3, 1) <= d <= date(2020, 6, 30):   return 0.15
    if date(2020, 7, 1) <= d <= date(2021, 6, 30):   return 0.45
    if date(2021, 7, 1) <= d <= date(2021, 12, 31):  return 0.75
    return 1.0

rows = []
start = date(2020, 1, 1)
end = date(2025, 12, 31)

for name, cfg in ATTRACTIONS.items():
    base_visitors = cfg["base"]
    place_type = cfg["type"]
    current = start
    while current <= end:
        weekday = current.weekday()
        is_weekend = weekday >= 5
        is_holiday = (current.month, current.day) in HOLIDAYS
        season = get_season(current.month)
        weather = np.random.choice(WEATHER_BY_SEASON[season])

        day_coef = 1.4 if weekday == 5 else 1.3 if weekday == 6 else 1.0
        season_coef = 1.3 if season == "summer" else 0.7 if season == "winter" else 1.0
        holiday_coef = 1.5 if is_holiday else 1.0
        weather_coef = WEATHER_IMPACT[place_type][weather]
        days_from_start = (current - start).days
        trend_coef = 1.0 + 0.03 * (days_from_start / 365)
        covid_coef = covid_factor(current)
        noise = np.random.normal(1.0, 0.07)

        visitors = int(
            base_visitors * day_coef * season_coef * holiday_coef
            * weather_coef * trend_coef * covid_coef * noise
        )
        visitors = max(visitors, 10)

        rows.append({
            "attraction": name,
            "date": current,
            "day_of_week": current.strftime("%A"),
            "is_weekend": is_weekend,
            "is_holiday": is_holiday,
            "season": season,
            "weather": weather,
            "base_popularity": base_visitors,
            "place_type": place_type,
            "is_covid_period": covid_coef < 1.0,
            "visitors": visitors,
        })
        current += timedelta(days=1)

df = pd.DataFrame(rows)
df.to_csv("attractions_data.csv", index=False)
```

```python
df = pd.read_csv("attractions_data.csv")
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

df["month"] = df["date"].dt.month
df["weekday_num"] = df["date"].dt.weekday
df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)
df["day_of_year"] = df["date"].dt.dayofyear
df["quarter"] = df["date"].dt.quarter
df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
df["weekday_sin"] = np.sin(2 * np.pi * df["weekday_num"] / 7)
df["weekday_cos"] = np.cos(2 * np.pi * df["weekday_num"] / 7)

df = df.sort_values(["attraction", "date"]).reset_index(drop=True)
for lag in [1, 7, 14]:
    df[f"lag_{lag}"] = df.groupby("attraction")["visitors"].shift(lag)

df["rolling_mean_7"] = (
    df.groupby("attraction")["visitors"].shift(1).rolling(7).mean()
    .reset_index(0, drop=True)
)
df["rolling_mean_14"] = (
    df.groupby("attraction")["visitors"].shift(1).rolling(14).mean()
    .reset_index(0, drop=True)
)
df = df.dropna().reset_index(drop=True)

split_date_1 = df["date"].quantile(0.70)
split_date_2 = df["date"].quantile(0.85)

train = df[df["date"] <= split_date_1].copy()
val = df[(df["date"] > split_date_1) & (df["date"] <= split_date_2)].copy()
test = df[df["date"] > split_date_2].copy()

FEATURES = [
    "attraction", "day_of_week", "weather", "place_type", "season",
    "month", "weekday_num", "week_of_year", "day_of_year", "quarter",
    "is_weekend", "is_holiday", "is_covid_period",
    "month_sin", "month_cos", "weekday_sin", "weekday_cos",
    "lag_1", "lag_7", "lag_14",
    "rolling_mean_7", "rolling_mean_14",
]
CAT_FEATURES = ["attraction", "day_of_week", "weather", "place_type", "season"]
TARGET = "visitors"

X_train, y_train = train[FEATURES], train[TARGET]
X_val, y_val = val[FEATURES], val[TARGET]
X_test, y_test = test[FEATURES], test[TARGET]

y_train_log = np.log1p(y_train)
y_val_log = np.log1p(y_val)

X_train_cat = X_train.copy()
X_val_cat = X_val.copy()
X_test_cat = X_test.copy()
for col in CAT_FEATURES:
    X_train_cat[col] = X_train_cat[col].astype("category")
    X_val_cat[col] = X_val_cat[col].astype("category")
    X_test_cat[col] = X_test_cat[col].astype("category")
```

```python
def evaluate(y_true, y_pred, name, train_time):
    return {
        "model": name,
        "MAPE (%)": mean_absolute_percentage_error(y_true, y_pred) * 100,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R²": r2_score(y_true, y_pred),
        "Время обучения (с)": train_time,
    }

def predict_original(model, X, use_log=True):
    pred = model.predict(X)
    if use_log:
        pred = np.expm1(pred)
    return np.clip(pred, 0, None)
```

```python
results = []

start = time.time()
cat_model = CatBoostRegressor(
    iterations=5000, learning_rate=0.03, depth=6,
    loss_function="RMSE", early_stopping_rounds=200,
    random_seed=42, l2_leaf_reg=3, bagging_temperature=0.5, verbose=0,
)
cat_model.fit(X_train, y_train_log, cat_features=CAT_FEATURES,
              eval_set=(X_val, y_val_log), verbose=0)
cat_time = time.time() - start
cat_pred = predict_original(cat_model, X_test)
results.append(evaluate(y_test, cat_pred, "CatBoost", cat_time))
cat_model.save_model("catboost_model.cbm")

start = time.time()
xgb_model = XGBRegressor(
    n_estimators=2000, learning_rate=0.03, max_depth=6,
    subsample=0.8, colsample_bytree=0.8, reg_alpha=0.1, reg_lambda=0.1,
    early_stopping_rounds=100, random_state=42,
    enable_categorical=True, tree_method="hist", verbosity=0,
)
xgb_model.fit(X_train_cat, y_train_log,
              eval_set=[(X_val_cat, y_val_log)], verbose=False)
xgb_time = time.time() - start
xgb_pred = predict_original(xgb_model, X_test_cat)
results.append(evaluate(y_test, xgb_pred, "XGBoost", xgb_time))

start = time.time()
lgb_model = lgb.LGBMRegressor(
    n_estimators=2000, learning_rate=0.03, num_leaves=32, max_depth=6,
    subsample=0.8, colsample_bytree=0.8, reg_alpha=0.1, reg_lambda=0.1,
    random_state=42, verbose=-1,
)
lgb_model.fit(X_train_cat, y_train_log,
              eval_set=[(X_val_cat, y_val_log)],
              callbacks=[lgb.early_stopping(100, verbose=False)])
lgb_time = time.time() - start
lgb_pred = predict_original(lgb_model, X_test_cat)
results.append(evaluate(y_test, lgb_pred, "LightGBM", lgb_time))

results_df = pd.DataFrame(results)
results_df.to_csv("results.csv", index=False)
print("ИТОГОВАЯ ТАБЛИЦА СРАВНЕНИЯ")
print(results_df.to_string(index=False))
```

```python
df_cv = df.sort_values("date").reset_index(drop=True)
all_categories = {
    col: df_cv[col].astype("category").cat.categories.tolist()
    for col in CAT_FEATURES
}

def to_unified_category(X, categories_dict):
    X = X.copy()
    for col, cats in categories_dict.items():
        X[col] = pd.Categorical(X[col], categories=cats)
    return X

tscv = TimeSeriesSplit(n_splits=10)
mape_per_fold = {"CatBoost": [], "XGBoost": [], "LightGBM": []}

for tr_idx, te_idx in tscv.split(df_cv):
    tr = df_cv.iloc[tr_idx]
    te = df_cv.iloc[te_idx]
    X_tr, y_tr = tr[FEATURES], tr[TARGET]
    X_te, y_te = te[FEATURES], te[TARGET]
    y_tr_log = np.log1p(y_tr)
    X_tr_cat = to_unified_category(X_tr, all_categories)
    X_te_cat = to_unified_category(X_te, all_categories)

    m = CatBoostRegressor(iterations=1000, learning_rate=0.05, depth=6,
                          random_seed=42, verbose=0)
    m.fit(X_tr, y_tr_log, cat_features=CAT_FEATURES)
    pred = np.expm1(m.predict(X_te))
    mape_per_fold["CatBoost"].append(
        mean_absolute_percentage_error(y_te, pred) * 100)

    m = XGBRegressor(n_estimators=1000, learning_rate=0.05, max_depth=6,
                     enable_categorical=True, tree_method="hist",
                     random_state=42, verbosity=0)
    m.fit(X_tr_cat, y_tr_log)
    pred = np.expm1(m.predict(X_te_cat))
    mape_per_fold["XGBoost"].append(
        mean_absolute_percentage_error(y_te, pred) * 100)

    m = lgb.LGBMRegressor(n_estimators=1000, learning_rate=0.05,
                          num_leaves=32, random_state=42, verbose=-1)
    m.fit(X_tr_cat, y_tr_log)
    pred = np.expm1(m.predict(X_te_cat))
    mape_per_fold["LightGBM"].append(
        mean_absolute_percentage_error(y_te, pred) * 100)

cv_summary = pd.DataFrame({
    "Модель": ["CatBoost", "XGBoost", "LightGBM"],
    "Средний MAPE": [np.mean(mape_per_fold[m]) for m in mape_per_fold],
    "Std": [np.std(mape_per_fold[m]) for m in mape_per_fold],
    "Мин": [np.min(mape_per_fold[m]) for m in mape_per_fold],
    "Макс": [np.max(mape_per_fold[m]) for m in mape_per_fold],
})
cv_summary.to_csv("cv_summary.csv", index=False)
print("\nКРОСС-ВАЛИДАЦИЯ (10 ФОЛДОВ)")
print(cv_summary.to_string(index=False))

print("\nWilcoxon signed-rank test:")
for a, b in [("CatBoost", "XGBoost"), ("CatBoost", "LightGBM"),
             ("XGBoost", "LightGBM")]:
    _, p = wilcoxon(mape_per_fold[a], mape_per_fold[b])
    print(f"  {a} vs {b}: p = {p:.4f}")

print("\nPaired t-test:")
for a, b in [("CatBoost", "XGBoost"), ("CatBoost", "LightGBM"),
             ("XGBoost", "LightGBM")]:
    _, p = ttest_rel(mape_per_fold[a], mape_per_fold[b])
    print(f"  {a} vs {b}: p = {p:.4f}")
```

```python
HORIZON = 16

def build_feature_row(date_, attraction, weather, history):
    hist = history.sort_values("date").reset_index(drop=True)
    visitors = hist["visitors"].values
    return pd.DataFrame([{
        "attraction": attraction,
        "day_of_week": date_.strftime("%A"),
        "weather": weather,
        "place_type": hist["place_type"].iloc[-1],
        "season": get_season(date_.month),
        "month": date_.month,
        "weekday_num": date_.weekday(),
        "week_of_year": date_.isocalendar()[1],
        "day_of_year": date_.dayofyear,
        "quarter": (date_.month - 1) // 3 + 1,
        "is_weekend": date_.weekday() >= 5,
        "is_holiday": (date_.month, date_.day) in HOLIDAYS,
        "is_covid_period": False,
        "month_sin": np.sin(2 * np.pi * date_.month / 12),
        "month_cos": np.cos(2 * np.pi * date_.month / 12),
        "weekday_sin": np.sin(2 * np.pi * date_.weekday() / 7),
        "weekday_cos": np.cos(2 * np.pi * date_.weekday() / 7),
        "lag_1": visitors[-1],
        "lag_7": visitors[-7],
        "lag_14": visitors[-14],
        "rolling_mean_7": visitors[-7:].mean(),
        "rolling_mean_14": visitors[-14:].mean(),
    }])

def forecast_recursive(model, history, attraction, weather_forecast, horizon=HORIZON):
    """Рекурсивный прогноз: прогноз шага t добавляется в ряд и используется для шага t+1."""
    history = history.copy()
    last_date = history["date"].max()
    predictions = []
    for step in range(horizon):
        d = last_date + timedelta(days=step + 1)
        weather = weather_forecast[step]
        X_row = build_feature_row(d, attraction, weather, history)
        for col in CAT_FEATURES:
            X_row[col] = X_row[col].astype("category")
        pred = float(np.expm1(model.predict(X_row[FEATURES])[0]))
        pred = max(pred, 0)

        new_row = history.iloc[-1:].copy()
        new_row["date"] = d
        new_row["visitors"] = pred
        new_row["weather"] = weather
        new_row["day_of_week"] = d.strftime("%A")
        history = pd.concat([history, new_row], ignore_index=True)
        predictions.append({"date": d, "predicted_visitors": pred})
    return pd.DataFrame(predictions)

attraction = "Attraction_00"
history = df[df["attraction"] == attraction].tail(14).copy()
weather_forecast = ["Clear", "Cloudy", "Rain"] * 5 + ["Clear"]

forecast_df = forecast_recursive(cat_model, history, attraction, weather_forecast)
forecast_df.to_csv(f"forecast_{attraction}.csv", index=False)
print(f"\nРЕКУРСИВНЫЙ ПРОГНОЗ НА {HORIZON} ДНЕЙ ({attraction})")
print(forecast_df.to_string(index=False))
```
